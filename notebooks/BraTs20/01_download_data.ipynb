## Import & setup path (cực sạch)

````python
from pathlib import Path
import os

# Root project (Drive)
PROJECT_ROOT = Path("/content/drive/MyDrive/NCKH/nnUnet")

# Data folders
DATA_DIR = PROJECT_ROOT / "data"
RAW_EXTERNAL_DIR = DATA_DIR / "raw_external"

# Create folders if not exist
RAW_EXTERNAL_DIR.mkdir(parents=True, exist_ok=True)

print("📁 Project root:", PROJECT_ROOT)
print("📁 Raw external data dir:", RAW_EXTERNAL_DIR)
````

## Download dataset từ Kaggle

````python
import kagglehub

# Kaggle dataset ID
KAGGLE_DATASET = "awsaf49/brats20-dataset-training-validation"

print("⬇️ Downloading dataset from Kaggle...")
download_path = kagglehub.dataset_download(KAGGLE_DATASET)

print("✅ Dataset downloaded to:")
print(download_path)
````

## Copy dataset vào `data/raw_external`

````python
import shutil

download_path = Path(download_path)

target_dir = RAW_EXTERNAL_DIR / "brats20"

if target_dir.exists():
    print("⚠️ Dataset already exists, skip copying.")
else:
    print(f"📦 Copying dataset to {target_dir} ...")
    shutil.copytree(download_path, target_dir)
    print("✅ Copy completed.")
````

Sau cell này, chỉ làm việc với: `data/raw_external/brats20/`

## Quick sanity check
