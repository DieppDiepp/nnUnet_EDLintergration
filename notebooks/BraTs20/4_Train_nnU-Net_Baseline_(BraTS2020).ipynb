## 6. Tính nhãn WT (Whole) TC (Core) ET (Enhancing)

```python
import os
import numpy as np
import nibabel as nib
import pandas as pd
import json
from multiprocessing import Pool

# ================= CẤU HÌNH =================
# Đường dẫn folder chứa nhãn gốc (Ground Truth) - labelsTr
GT_FOLDER = "/content/nnUNet_raw/Dataset101_BraTS2020/labelsTr"

# Đường dẫn folder chứa kết quả dự đoán của nnU-Net (Validation Results)
# Lưu ý: Tìm folder validation_raw hoặc tương tự trong kết quả train
# Thường là: .../nnUNetTrainer_50epochs__nnUNetPlans__3d_fullres/fold_0/validation
# NHƯNG vì bạn đã chạy find_best_configuration, nó đã gộp kết quả 5 fold vào folder này:
PRED_FOLDER = "/content/nnUNet_results/Dataset101_BraTS2020/nnUNetTrainer_50epochs__nnUNetPlans__3d_fullres/crossval_results_folds_0_1_2_3_4"

# OUTPUT_FILE = "BraTS_Final_Metrics.csv"
OUTPUT_FILE = os.path.join(PRED_FOLDER, "BraTS_EDL_Metrics.csv")

# ================= HÀM TÍNH TOÁN =================

def compute_dice(pred_mask, gt_mask):
    """Tính Dice Score cho Binary Mask"""
    # Dice = 2 * (A giao B) / (|A| + |B|)
    intersection = np.sum(pred_mask * gt_mask)
    sum_pixels = np.sum(pred_mask) + np.sum(gt_mask)

    if sum_pixels == 0:
        return 1.0 # Cả 2 đều không có gì -> Đúng tuyệt đối
    return 2.0 * intersection / sum_pixels

def process_case(case_id):
    try:
        # 1. Đọc file
        # Tên file nnU-Net thường giữ nguyên hoặc thêm đuôi, check lại format tên file
        # Ví dụ: BraTS20_Training_001.nii.gz

        gt_path = os.path.join(GT_FOLDER, f"{case_id}.nii.gz")
        pred_path = os.path.join(PRED_FOLDER, f"{case_id}.nii.gz")

        if not os.path.exists(gt_path) or not os.path.exists(pred_path):
            return None

        gt_nii = nib.load(gt_path)
        pred_nii = nib.load(pred_path)

        gt_data = gt_nii.get_fdata().astype(np.uint8)
        pred_data = pred_nii.get_fdata().astype(np.uint8)

        # 2. Định nghĩa các vùng (Regions) theo BraTS 2020
        # Label 1: NCR/NET
        # Label 2: ED
        # Label 3: ET

        # --- ET (Enhancing Tumor): Label 3 ---
        gt_ET = (gt_data == 3)
        pred_ET = (pred_data == 3)

        # --- TC (Tumor Core): Label 1 + 3 ---
        gt_TC = np.logical_or(gt_data == 1, gt_data == 3)
        pred_TC = np.logical_or(pred_data == 1, pred_data == 3)

        # --- WT (Whole Tumor): Label 1 + 2 + 3 ---
        gt_WT = (gt_data > 0) # Mọi cái khác 0 đều là u
        pred_WT = (pred_data > 0)

        # 3. Tính Dice
        dice_ET = compute_dice(pred_ET, gt_ET)
        dice_TC = compute_dice(pred_TC, gt_TC)
        dice_WT = compute_dice(pred_WT, gt_WT)

        return {
            "Case_ID": case_id,
            "Dice_ET": dice_ET,
            "Dice_TC": dice_TC,
            "Dice_WT": dice_WT
        }
    except Exception as e:
        print(f"Error processing {case_id}: {e}")
        return None

# ================= MAIN =================
if __name__ == "__main__":
    print("Đang tính toán metric BraTS (WT, TC, ET)...")

    # Lấy danh sách file từ folder dự đoán (chỉ lấy file .nii.gz)
    files = [f for f in os.listdir(PRED_FOLDER) if f.endswith('.nii.gz')]
    case_ids = [f.replace('.nii.gz', '') for f in files]

    print(f"Dataset size: {len(case_ids)} cases")

    results = []
    # Chạy loop thường hoặc multiprocessing (Colab có 2 core thì chạy loop cũng nhanh)
    for i, cid in enumerate(case_ids):
        res = process_case(cid)
        if res:
            results.append(res)
        if i % 20 == 0:
            print(f"Processed {i}/{len(case_ids)}")

    # Tạo DataFrame
    df = pd.DataFrame(results)

    # Tính trung bình
    summary = {
        "Mean_Dice_ET": df["Dice_ET"].mean(),
        "Mean_Dice_TC": df["Dice_TC"].mean(),
        "Mean_Dice_WT": df["Dice_WT"].mean()
    }

    print("\n" + "="*30)
    print("KẾT QUẢ CUỐI CÙNG (BASELINE)")
    print("="*30)
    print(df.describe())
    print("-" * 30)
    print(f"Mean Dice WT: {summary['Mean_Dice_WT']:.4f}")
    print(f"Mean Dice TC: {summary['Mean_Dice_TC']:.4f}")
    print(f"Mean Dice ET: {summary['Mean_Dice_ET']:.4f}")

    # Lưu file
    df.to_csv(OUTPUT_FILE, index=False)
    print(f"\nĐã lưu kết quả chi tiết vào: {OUTPUT_FILE}")
```

## 7. Sao lưu & Nén kết quả (Safe Backup)

```python
import os
import shutil
import subprocess

# ================= CẤU HÌNH ĐƯỜNG DẪN =================
colab_results_base = "/content/nnUNet_results"
dataset_name = "Dataset101_BraTS2020"
colab_dataset_path = os.path.join(colab_results_base, dataset_name)

# Pattern tên folder Trainer Baseline
target_trainer_prefix = "nnUNetTrainer_50epochs"

# Đường dẫn đích trên Google Drive
drive_root = "/content/drive/MyDrive/NCKH/nnUnet/data/nnUNet_results"
drive_dataset_path = os.path.join(drive_root, dataset_name)

# ================= XỬ LÝ LOGIC =================

def backup_results():
    print(f"Bắt đầu cập nhật Backup cho: {dataset_name}")

    # --- BƯỚC 1: TÌM FOLDER TRAINER ---
    available_folders = [f for f in os.listdir(colab_dataset_path)
                         if os.path.isdir(os.path.join(colab_dataset_path, f))]

    trainer_folder_name = None
    for f in available_folders:
        if target_trainer_prefix in f:
            trainer_folder_name = f
            break

    if not trainer_folder_name:
        print(f"Không tìm thấy folder trainer nào khớp '{target_trainer_prefix}'")
        return

    print(f"Target Trainer: {trainer_folder_name}")
    print("Lưu ý: File CSV metric của bạn đã nằm trong folder con 'crossval_results...', script sẽ tự động backup nó.")

    # --- BƯỚC 2: ĐỒNG BỘ SANG DRIVE (RSYNC) ---
    print("\n[1/2] Đồng bộ sang Drive (Chỉ update file mới)...")
    os.makedirs(drive_dataset_path, exist_ok=True)

    # Rsync sẽ quét toàn bộ folder trainer.
    # Vì file CSV của bạn nằm trong folder con của trainer này nên nó sẽ ĐƯỢC copy theo.
    src_trainer = os.path.join(colab_dataset_path, trainer_folder_name)

    # Lệnh rsync
    subprocess.run(["rsync", "-a", "--info=progress2", src_trainer, drive_dataset_path])

    # Copy các file json/txt ở root dataset (inference info...)
    files_to_copy = [f for f in os.listdir(colab_dataset_path)
                     if f.endswith(('.json', '.txt', '.pkl')) and os.path.isfile(os.path.join(colab_dataset_path, f))]

    for f in files_to_copy:
        src_f = os.path.join(colab_dataset_path, f)
        subprocess.run(["cp", src_f, drive_dataset_path])

    print(f"   -> Đồng bộ hoàn tất.")

    # --- BƯỚC 3: CẬP NHẬT FILE ZIP ---
    print("\n[2/2] Cập nhật file Zip...")

    zip_name = f"{dataset_name}_Baseline_50epochs.zip"
    zip_path_drive = os.path.join(drive_dataset_path, zip_name)

    cwd = os.getcwd()
    os.chdir(colab_results_base)

    # List cần zip
    paths_to_zip = [
        os.path.join(dataset_name, trainer_folder_name)
    ]
    for f in files_to_copy:
        paths_to_zip.append(os.path.join(dataset_name, f))

    try:
        # Zip sẽ tự động ghi đè/update file cũ trên Drive
        cmd = ["zip", "-r", "-q", zip_path_drive] + paths_to_zip
        subprocess.run(cmd, check=True)
        print(f"Đã cập nhật file nén tại: {zip_path_drive}")
        print("   (File CSV mới đã được bao gồm trong file zip)")
    except Exception as e:
        print(f"Lỗi khi zip: {e}")
    finally:
        os.chdir(cwd)
```
