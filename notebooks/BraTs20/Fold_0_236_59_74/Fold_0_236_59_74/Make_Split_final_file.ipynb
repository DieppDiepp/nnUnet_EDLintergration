```python
import os
import json
import numpy as np
from sklearn.model_selection import KFold

# --- ĐƯỜNG DẪN FILE ---
base_dir = "/content/drive/MyDrive/NCKH/nnUnet/data/nnUNet_preprocessed/Dataset101_BraTS2020"
backup_split_path = os.path.join(base_dir, "splits_final_5fold_backup.json")
new_split_path = os.path.join(base_dir, "splits_final.json")

# Đổi thành file JSON
test_dir = "/content/drive/MyDrive/NCKH/nnUnet/data/experiments"
os.makedirs(test_dir, exist_ok=True)
test_file_path = os.path.join(test_dir, "fixed_test.json")

# 1. ĐỌC FILE SPLIT CŨ
print("Đang đọc file split backup...")
with open(backup_split_path, 'r') as f:
    old_splits = json.load(f)

# 2. TẠO TẬP TEST TỪ TẤT CẢ CÁC FOLD CŨ
test_splits = {}
for i, split_data in enumerate(old_splits):
    test_splits[f"fold_{i}"] = split_data['val']

print(f"Đã bóc tách Test Set của {len(test_splits)} folds cũ.")
print(f"   -> Riêng Fold 0 có {len(test_splits['fold_0'])} ca Test.")

# 3. LẤY DANH SÁCH 295 CA CÒN LẠI (Tập Train của Fold 0 cũ)
cv_cases = old_splits[0]['train']
print(f"\nSố lượng ca bệnh đưa vào 5-Fold CV mới: {len(cv_cases)} ca.")

# 4. CHIA 5 FOLDS TRÊN 295 CA NÀY
cv_cases = np.array(cv_cases)
kf = KFold(n_splits=5, shuffle=True, random_state=42)

new_splits = []
for fold_idx, (train_idx, val_idx) in enumerate(kf.split(cv_cases)):
    fold_train = cv_cases[train_idx].tolist()
    fold_val = cv_cases[val_idx].tolist()

    new_splits.append({
        "train": fold_train,
        "val": fold_val
    })
    print(f"   -> New Fold {fold_idx}: Train {len(fold_train)} ca | Val {len(fold_val)} ca")

# 5. LƯU CÁC FILE MỚI (Đều là JSON)
with open(new_split_path, 'w') as f:
    json.dump(new_splits, f, indent=4)

with open(test_file_path, 'w') as f:
    json.dump(test_splits, f, indent=4)

print(f"\nTHÀNH CÔNG! Đã lưu file:")
print(f" - {new_split_path}")
print(f" - {test_file_path}")
```
